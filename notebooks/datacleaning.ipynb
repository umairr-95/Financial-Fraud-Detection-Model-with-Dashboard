```python
#importing libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sn
```

```python
#read csv file

df = pd.read_csv("D:/Fraud Detection model project 2/Raw_dataset/financial_fraud_detection_dataset.csv")
```

```python
df.info()
```

```python
df.describe()
```

```python
# missing values in the dataset
df.isnull().sum()
```

```python
# missing percentage in the dataset
missing_percentage = df.isnull().sum() / len(df) * 100
print(missing_percentage)
```

```python
missing_summary = pd.DataFrame({'Missing Values': df.isnull().sum(), 'Percentage': missing_percentage}) 
missing_summary
```

```python
# checking duplicate transactions in the dataset
duplicate_transactions = df.duplicated().sum()
print(f"Number of duplicate transactions: {duplicate_transactions}")
```

```python
# checking duplicates exist in the dataset
df = df.drop_duplicates()
print(f"Shape after removing duplicates: {df.shape}")
```

```python
# checking columns names in dataset
df.columns.to_list()
```

```python
# removing unnecessary columns from the dataset
print(df['Unnamed: 14'].isnull().all())
```

```python
# now we can remove the column 'Unnamed: 14' from the dataset as it contains all null values.
df = df.drop(['Unnamed: 14'], axis=1)
```

```python
print(df.shape)
print(df.columns.tolist())
```

```python
#check datatypes
print(df.dtypes)
```

```python
df["Transaction_Date"]= pd.to_datetime(df["Transaction_Date"], format= "%d-%m-%Y %H:%M:%S", errors= "coerce")
```

```python
print(df["Transaction_Date"].dtypes)
```

```python
#checking for date conversion created problems
invalid_dates = df[df["Transaction_Date"].isnull()]
print("invalid_dates", invalid_dates)
```

```python
# checking numeric columns in the dataset
numeric_columns = df.select_dtypes(include=[np.int64, np.float64]).columns.tolist()
print("Numeric columns:", numeric_columns)
```

```python
df[numeric_columns].describe()
```

```python
# checking categorical columns in the dataset
categorical_columns = df.select_dtypes(include=[object]).columns
print(categorical_columns) 

for col in categorical_columns:
    print(f"Column: {col}")
    print(df[col].value_counts())
    print("\n")
```

```python
# checking the target for fraud detection in the dataset
print(df['Fraudulent'].value_counts())

print(df['Fraudulent'].value_counts(normalize=True) * 100)
```

```python
# checking for all null values, missing values, and duplicates in the dataset

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nData types:")
print(df.dtypes)
```

```python
# creating a new csv file for cleaned dataset

output_path = "D:\\Fraud Detection model project 2\\Raw_dataset\\financial_fraud_detection_cleaned.csv"

df.to_csv(output_path, index=False)

print("Cleaned dataset saved successfully.")
```
